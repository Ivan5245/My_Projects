# 1

```python
a=[1,2,3,4,5]
k=2

n=len(a)
if(n%k==0):
  q=n//k
  k1=k
else:
  q=n
  k1=1

for i1 in range(k1):
  next=i1
  prev=a[i1]
  for _ in range(q):
    next=(next+k)%n
    cur=a[next]
    a[next]=prev
    prev=cur


print(a)
```

Данное решение использует линейное время, так как состоит из двойного цикла for, где произведение пределов итерирования равно n. Выделяемая память константна, так как не происходит выделения массивов.

# 2

```python
a=[1,2,3,4,5,6]
b=[2,4,6,8]
c=[]

j=0
for i in range(len(a)):
  while(b[j]<=a[i] and j<len(b)):
    c.append(b[j])
    j+=1
  c.append(a[i])
while(j!=len(b)):
  c.append(b[j])
  j+=1
print(c)
```

Данный алгоритм строит массив c из всех элементов массивов a и b за счет двух указателей. Отсортированность достигается за счет того, что перед добавлением следующего элемента из a добавляются все элементы из b не превосходящие элемент a

# 3

```python
a=[1,2,3,4,5,6]+[0]*4
b=[2,4,6,8]

l1=len(a)-len(b)
l2=len(b)
e=l1-1
w=len(a)-1
for i in range(l2-1,-1,-1):
  while(a[e]>=b[i]):
    a[w]=a[e]
    w-=1
    e-=1
  a[w]=b[i]
  w-=1
a
```

Заполнение с конца позволяет избежать перезаписи данных, так как для того, чтобы дойти до k-й с конца ячейки первого массива для записи мы должны уже переписать все ячейки после k-й, следовательно, либо k-я ячейка уже была переписана, либо она должна быть переписана в себя, в любом случае мы уже воспользовались информацией из этой ячейки.  

# 4

```python
a=[1,0,0,0,1,0,0,1,0,1,1,1]

j=len(a)-1
i=0
while(i<j):
  while(a[j]==1 and i<j):
    j-=1
  while(a[i]==0 and i<j):
    i+=1
  a[i]=0
  a[j]=1
a
```

Алгоритм имеет линейную сложность, так как состоит из перемещения двух указателей, посещающих каждую ячейку массива ровно по одному разу.

```python
a=[0,1,2,2,2,0,1,2,0,1,2,2,0,1,2,0,2,2]

l=0
r=len(a)-1
while(a[l]==0):
  l+=1
while(a[r]==2):
  r-=1
m=l
while(m<r):
  if(a[m]==1):
    m+=1
  if(a[m]==0):
    a[m]=a[l]
    a[l]=0
    while(a[l]==0):
      l+=1
  if(a[m]==2):
    a[m]=a[r]
    a[r]=2
    while(a[r]==2):
      r-=1
a
```

l-индекс первой в массиве ячейки со значением не 0
r-индекс последней в массиве ячейки со значением не 2
m-индекс первой в массиве ячейки после блока 0 и блока 1, скрепленных последовательно

```python
a= [3, 2, 4, 1, 11, 8, 9,10,12,8,1,6,7,3]

l=0
r=0
while(l<len(a) and r<len(a)):
  while(a[l]%2==0):
    l+=1
  if(r<=l):
    r=l
  while(r<len(a) and a[r]%2==1):
    r+=1
  if(r<len(a)):
    c=a[r]
    a[r]=a[l]
    a[l]=c
a
```

Порядок сохраняется благодаря тому, что между l и r никогда нет четных элементов массива, следовательно, при замене порядок четных элементов не меняется

# 7

```python
a=[2,4,5,2,6,0,0,7,0,4,3,0,4,2,0,2,3,0,1]

l=0
r=0
while(l<len(a) and r<len(a)):
  while(a[l]!=0):
    l+=1
  if(r<l):
    r=l
  while(r<len(a) and a[r]==0):
    r+=1
  if(r<len(a)):
    a[l]=a[r]
    a[r]=0
    l+=1
a
```

Решение работает за O(n) так как в циклах происходит не более 2n итераций (для r и l), оно является стабильным, так как при замене элементов l и r между ними находятся только нули, следовательно исходный порядок ненулевых элементов сохраняется.
