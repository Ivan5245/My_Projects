```python
a=[1,2,3,4,5]
k=2

n=len(a)
if(n%k==0):
  q=n//k
  k1=k
else:
  q=n
  k1=1

for i1 in range(k1):
  next=i1
  prev=a[i1]
  for _ in range(q):
    next=(next+k)%n
    cur=a[next]
    a[next]=prev
    prev=cur


print(a)
```

Данное решение использует линейное время, так как состоит из двойного цикла for, где произведение пределов итерирования равно n. Выделяемая память константна, так как не происходит выделения массивов.
